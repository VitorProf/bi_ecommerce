# Versionamento da Ingestão de Logística no Git

Após finalizar o processo de ingestão dos arquivos de logística no Azure Data Lake, podemos registrar essa versão do projeto no Git.

O fluxo desenvolvido nesta etapa foi:

**CSV → Landing → Validação Técnica → Bronze → Registro de Processamento**



## 1. Verificar o status do projeto

Antes de adicionar os arquivos ao Git, devemos verificar quais arquivos foram criados ou modificados.

```bash
git status
```

O comando `git status` permite visualizar:

- arquivos modificados;
- arquivos novos;
- arquivos preparados para commit;
- arquivos ainda não adicionados ao Git.

> **Importante:** o arquivo `.env` contém a credencial de acesso ao Azure e não deve ser enviado ao Git.



## 2. Adicionar as alterações

Para preparar as alterações que serão incluídas no próximo commit, utilizamos:

```bash
git add .
```

O comando `git add .` adiciona as alterações à **Staging Area**, que representa a área de preparação para o próximo commit.

O ponto `.` representa o diretório atual. Portanto, o Git considera os arquivos novos, modificados e removidos do diretório atual e de suas subpastas.

Após adicionar os arquivos, podemos verificar novamente:

```bash
git status
```

Os arquivos preparados deverão aparecer na seção:

```text
Changes to be committed
```

> Arquivos configurados no `.gitignore`, como `.env`, `.venv/` e `.idea/`, não serão adicionados pelo `git add .`, desde que ainda não estejam sendo rastreados pelo Git.



## 3. Registrar a versão do projeto

Depois de preparar os arquivos, criamos um **commit** para registrar essa versão no histórico do projeto.

```bash
git commit -m "Implementa ingestao da logistica da Landing para Bronze"
```

A opção `-m` permite adicionar uma mensagem descrevendo o que foi desenvolvido nessa versão.

Neste caso, o commit registra a conclusão da etapa:

**Landing → Validação → Bronze**



## 4. Verificar o histórico

Para visualizar os commits realizados:

```bash
git log --oneline
```

Esse comando apresenta o histórico de forma resumida, mostrando o identificador e a mensagem de cada commit.

Exemplo:

```text
a8f24c1 Implementa ingestao da logistica da Landing para Bronze
```
